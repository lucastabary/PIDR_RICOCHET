```python
#Imports
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D

data = pd.read_csv("./Data/CIRAD_Sorghum2023_SPIR_all_2 (4).csv",sep=";")
```

```python
#Control plantes
Control_plantes = data.drop(data[data.Rep>=2].index)
Control_plantes = Control_plantes.loc[data["Traitement"]=="C",:]
ID_count = Control_plantes["ID"].value_counts()
ID_indexes = ID_count.index.to_list() # range dans une liste la liste des index pour lequel la plante a le plus de valeur

Control_plante = Control_plantes[Control_plantes.ID == ID_indexes[0]]

##ploting
wavelength = [i for i in range(350,2501)]
x_data_control= Control_plante.loc[:,"350":"2500"].to_numpy()

plt.figure(figsize=(12,6))
plt.xlabel("Wavelength")
plt.ylabel("Absorption")
fill = plt.subplot()

import matplotlib.cm as cm
num_curves = len(x_data_control)
colors = cm.get_cmap('coolwarm', num_curves)
for idx, k in enumerate(x_data_control):
    fill.plot(wavelength, k, color=colors(idx))
```

```python
#SSO
SS0_plantes = data.drop(data[data.Rep>=2].index)
SS0_plantes = SS0_plantes.loc[data["Traitement"]=="SS0",:]
ID_count = SS0_plantes["ID"].value_counts()
ID_indexes = ID_count.index.to_list() # range dans une liste la liste des index pour lequel la plante a le plus de valeur

SS0_plante = SS0_plantes[SS0_plantes.ID == ID_indexes[0]]

##ploting
wavelength = [i for i in range(350,2501)]
x_data_control= SS0_plante.loc[:,"350":"2500"].to_numpy()

plt.figure(figsize=(12,6))
plt.xlabel("Wavelength")
plt.ylabel("Absorption")
fill = plt.subplot()

import matplotlib.cm as cm
num_curves = len(x_data_control)
colors = cm.get_cmap('coolwarm', num_curves)
for idx, k in enumerate(x_data_control):
    fill.plot(wavelength, k, color=colors(idx))

# utiliser la date de chaque mesure comme label (détection automatique de la colonne date)
date_col = None
for c in SS0_plante.columns:
    if any(x in c.lower() for x in ['date','jour','day','timestamp']):
        date_col = c
        break
if date_col is None:
    if 'Rep' in SS0_plante.columns:
        dates = SS0_plante['Rep'].astype(str).tolist()
    else:
        dates = [str(i) for i in range(num_curves)]
else:
    dates = SS0_plante[date_col].astype(str).tolist()
legend_elements = [Line2D([0], [0], color=colors(i), lw=2, label=dates[i]) for i in range(min(num_curves, len(dates)))]
fill.legend(handles=legend_elements, loc='upper right')
```
