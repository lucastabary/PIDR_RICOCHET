```python
#Imports
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


data = pd.read_csv("./Data/CIRAD_Sorghum2023_SPIR_all_2 (4).csv",sep=";")
```

```python
#Control plantes
Control_plantes = data.drop(data[data.Rep>=2].index)
Control_plantes = Control_plantes.loc[data["Traitement"]=="C",:]
ID_count = Control_plantes["ID"].value_counts()
ID_indexes = ID_count.index.to_list() # range dans une liste la liste des index pour lequel la plante a le plus de valeur

Control_plante = Control_plantes[Control_plantes.ID == ID_indexes[0]]

##ploting
wavelength = [i for i in range(350,2501)]
x_data_control= Control_plante.loc[:,"350":"2500"].to_numpy()

plt.figure(figsize=(12,6))
plt.xlabel("Wavelength")
plt.ylabel("Absorption")
fill = plt.subplot()

import matplotlib.cm as cm
num_curves = len(x_data_control)
colors = cm.get_cmap('coolwarm', num_curves)
for idx, k in enumerate(x_data_control):
    fill.plot(wavelength, k, color=colors(idx))
```
